```python
# Q7. Sum of weights
# 1. Asia
car_fuel_eff_df_asia = car_fuel_eff_df[car_fuel_eff_df["origin"] == "Asia"]
# 2. Columns
car_fuel_eff_df_asia = car_fuel_eff_df_asia[["vehicle_weight","model_year"]]
# 3. Read 7 rows
car_fuel_eff_df_asia.head(7)
```

```python
# 4. Numpy array
import numpy as np
X = np.array(car_fuel_eff_df_asia.head(7))
X
```

```python
# 5. Matrix-Matrix multiplication
XTX = X.T.dot(X)
XTX
```

```python
# 6. Inverse
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# 7. Array y
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
# 8. Result w
w = XTX_inv.dot(X.T).dot(y)
w
```

```python
# 9. Sum of result elements
sum(w)
```
